```python
import pandas as pd
import numpy as np
np.random.seed(42)
```

```python
data = {
    'total_area': np.random.uniform(low=30, high=150, size=100),
    'rooms_count': np.random.randint(low=1, high=5, size=100),
    'floor': np.random.randint(low=1, high=20, size=100),
    'district': ['Север', 'Восток', 'Запад', 'Юг'] * 25,
    'distance_from_center': np.random.uniform(low=1, high=20, size=100)
}
```

```python
data['prices'] = (
    data['total_area'] * 40000 +
    data['rooms_count'] * 10000 +
    (-data['floor'] * 1000) +
    (-data['distance_from_center'] * 500) +
    np.random.normal(loc=0, scale=5000, size=100)
)
```

```python
df = pd.DataFrame(data)
data_fram = df.round({'total_area':1, 'distance_from_center':1, 'prices':1})
data_fram.head()
```

```python
data_fram.to_csv("House.csv", index=False)
```
