```python
def get_continent(country):

    # Азия
    asia_countries = ['China', 'India', 'Japan', 'South Korea', 'Indonesia', 
                      'Russia', 'Saudi Arabia', 'Iran', 'Iraq', 'Turkey',
                      'Thailand', 'Vietnam', 'Philippines', 'Malaysia', 'Pakistan',
                      'Bangladesh', 'Afghanistan', 'Uzbekistan', 'Kazakhstan',
                      'North Korea', 'Sri Lanka', 'Myanmar', 'Yemen', 'Nepal',
                      'Taiwan', 'Israel', 'Hong Kong', 'Singapore', 'Jordan',
                      'United Arab Emirates', 'Qatar', 'Kuwait', 'Oman', 'Lebanon',
                      'Georgia', 'Azerbaijan', 'Armenia', 'Bahrain', 'Cyprus',
                      'Bhutan', 'Brunei', 'Cambodia', 'Laos', 'Mongolia',
                      'Tajikistan', 'Turkmenistan', 'Kyrgyzstan', 'Macao', 'Palestine',
                      'Timor', 'Maldives']
    
    # Европа
    europe_countries = ['Germany', 'United Kingdom', 'France', 'Italy', 'Spain',
                        'Poland', 'Netherlands', 'Belgium', 'Sweden', 'Austria',
                        'Switzerland', 'Norway', 'Ireland', 'Denmark', 'Finland',
                        'Portugal', 'Czech Republic', 'Greece', 'Romania', 'Hungary',
                        'Slovakia', 'Croatia', 'Belarus', 'Bulgaria', 'Serbia',
                        'Lithuania', 'Slovenia', 'Latvia', 'Estonia', 'Luxembourg',
                        'Bosnia and Herzegovina', 'Albania', 'North Macedonia',
                        'Moldova', 'Malta', 'Iceland', 'Montenegro', 'Andorra',
                        'Liechtenstein', 'Monaco', 'San Marino', 'Vatican',
                        'Czechoslovakia', 'Soviet Union', 'Yugoslavia']
    
    # Северная Америка
    north_america_countries = ['United States', 'Canada', 'Mexico', 'Guatemala',
                               'Cuba', 'Haiti', 'Dominican Republic', 'Honduras',
                               'Nicaragua', 'El Salvador', 'Costa Rica', 'Panama',
                               'Jamaica', 'Puerto Rico', 'Trinidad and Tobago',
                               'Bahamas', 'Barbados', 'Saint Lucia', 'Grenada',
                               'Belize', 'Antigua and Barbuda', 'Dominica',
                               'Saint Kitts and Nevis', 'Saint Vincent and the Grenadines']
    
    # Южная Америка
    south_america_countries = ['Brazil', 'Argentina', 'Colombia', 'Peru', 'Venezuela',
                               'Chile', 'Ecuador', 'Bolivia', 'Paraguay', 'Uruguay',
                               'Guyana', 'Suriname', 'French Guiana']
    
    # Африка
    africa_countries = ['Nigeria', 'Egypt', 'South Africa', 'Algeria', 'Morocco',
                        'Ethiopia', 'Kenya', 'Tanzania', 'Ghana', 'Democratic Republic of Congo',
                        'Uganda', 'Sudan', 'South Sudan', 'Angola', 'Mozambique',
                        'Madagascar', 'Cameroon', "Cote d'Ivoire", 'Niger', 'Burkina Faso',
                        'Mali', 'Malawi', 'Zambia', 'Senegal', 'Chad', 'Somalia',
                        'Zimbabwe', 'Guinea', 'Rwanda', 'Benin', 'Burundi', 'Tunisia',
                        'South Sudan', 'Togo', 'Sierra Leone', 'Libya', 'Congo',
                        'Central African Republic', 'Liberia', 'Mauritania', 'Eritrea',
                        'Namibia', 'Botswana', 'Gabon', 'Lesotho', 'Guinea-Bissau',
                        'Equatorial Guinea', 'Mauritius', 'Eswatini', 'Djibouti',
                        'Comoros', 'Cabo Verde', 'Sao Tome and Principe', 'Seychelles']
    
    # Океания
    oceania_countries = ['Australia', 'New Zealand', 'Papua New Guinea', 'Fiji',
                         'Solomon Islands', 'Vanuatu', 'New Caledonia', 'French Polynesia',
                         'Samoa', 'Kiribati', 'Micronesia', 'Tonga', 'Marshall Islands',
                         'Palau', 'Tuvalu', 'Nauru', 'Cook Islands', 'Niue']
    
    if country in asia_countries:
        return 'Asia'
    elif country in europe_countries:
        return 'Europe'
    elif country in north_america_countries:
        return 'North America'
    elif country in south_america_countries:
        return 'South America'
    elif country in africa_countries:
        return 'Africa'
    elif country in oceania_countries:
        return 'Oceania'
    else:
        return 'Other'
    

df['Continent'] = df['Entity'].apply(get_continent)

df.head(5)
```

```python
continent_totals = df.groupby('Continent')['Annual CO₂ emissions (tonnes )'].sum()
```

```python
print(f"Доминирующий континент по выбросам CO₂: {continent_totals.idxmax()}, Выбросы: {continent_totals.max()}")
```
